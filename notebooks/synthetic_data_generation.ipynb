Fill the csv file in data folder with symthetic data for testing and training the model using sionna

```python
import numpy as np
import pandas as pd
import sionna as sn
import os as path

# Set the random seed for reproducibility
np.random.seed(42)

# Define the number of samples to generate
num_samples = 15000

snr_db = np.random.uniform(0, 30, num_samples)  # SNR values between 0 and 30 dB
modulation_schemes = ['QPSK', '16QAM', '64QAM']
modulation = np.random.choice(modulation_schemes, num_samples)  # Randomly select modulation schemes
# Create a DataFrame to hold the synthetic data
data = pd.DataFrame({
    'SNR_dB': snr_db,
    'Modulation': modulation
})
# Save the DataFrame to a CSV file
data.to_csv('data/network_data.csv', index=False)
print("Synthetic data generated and saved to 'data/network_data.csv'")
```
